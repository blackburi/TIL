# 4가지 keyword
1. API
    * 프로그램 간 데이터를 전달하기 위해 정한 규칙
2. HTTP
    * Web에서 데이터를 주고받기 위한 프로토콜
    * https : http + 보안(security)
3. JSON
    * 데이터 전달 포맷
    * 장점 : HTML이나 XML보다 사람이 읽기 편하고 간단하게 python 객체로 변환할 수 있음
4. XML
    * JSON보다 조금 ㅈ아황하지만 사람이 이해하기 쉬운 구조적인 포맷
    * python에서 기본적으로 제공되는 xml 패키지를 사용하여 XML 문서에 있는 엘리먼트를 탐색할 수 있음
    * pandas의 경우 `read_xml()`함수를 사용하여 데이터프레임으로 바꿀 수 있음

# 핵심 함수 및 매서드
* `json.dumps()` : python 객체 -> JSON 문자열 변환
* `json.loads()` : JSON 문자열 -> python 객체
* `pandas.read_json()` : JSON 문자열 -> pandas 시리즈 or 데이터프레임
* `xml.etree.ElementTree.fromstring()` : XML 문자열 -> xml.etree.ElementTree.Element 클래스 객체 반환
* `xml.etree.ElementTree.Element.findtext()` : 지정한 tag name과 맞는 첫 번째 자식 element의 text를 반환
* `xml.etree.ElementTree.Element.findall()` : 지정한 tag name과 맞는 모든 자식 element를 반환
* `requests.get()` : HTTP GET 방식으로 URL 호출 및 requests.Response 객체 반환
* `requests.Response.json()` : 응답받은 JSON 문자열 -> python 객체로 변환하여 반환

```python
# dictionary 생성
d = {"name" : "김태경"}
print(d)
print(type(d))
print(d["name"])
```

```python
# python 객체를 json 문자열로 변환
import json

# json.dumps()함수는 아스키 문자 외의 다른 문자를 16진수로 출력하기 때문에 한글이 제대로 보이지 않음
d_str = json.dumps(d)
print(d_str)

# json.dumps()함수에서 한글을 보기 위해서 ensure_ascii 매개변수를 False로 설정
d_str = json.dumps(d, ensure_ascii=False)
print(d_str)
print(type(d_str))
```

```python
# json 문자열을 python 객체로 변환
d2 = json.loads(d_str)

print(d2)
print(type(d2))
```

```python
# 여러 항목을 가진 json 문자열을 python 객체로 변환
d3 = json.loads('{"name" : "김태경", "job" : ["개발자", "댄서"], "age" : 20}')

print(d3)
print(d3["name"])
print(type(d3["job"]))
print(d3["job"][1])
print(d3["age"])
print(type(d3))
```

```python
d4_str = """
[
    {"name" : "김태경1", "job" : "개발자", "age" : 20},
    {"name" : "김태경2", "job" : "댄서", "age" : 21}
]
"""

d4 = json.loads(d4_str)
print(d4[0]["name"])
```

```python
# json 문자열을 데이터프레임으로 변환하는 방법 1 - 문자열을 바로 데이터프레임으로 변환
from io import StringIO
import pandas as pd

pd.read_json(StringIO(d4_str))
```

```python
# json 문자열을 데이터프레임으로 변환하는 방법 2 - 문자열을 python 객체로 변환 후 데이터프레임으로 변환
pd.DataFrame(d4)
```

```python
# XML문자열을 python 객체로 변환

import xml.etree.ElementTree as et

# json pkg는 json 문자열을 python 객체로 변환
# fromstring() 함수가 반환하는 객체는 단순 python 객체가 아닌 ElementTree 모듈 아래에 정의된 Element 클래스의 객체

x_str = """
<info>
    <name>김태경</name>
    <job>개발자</job>
    <age>20</age>
</info>
"""

my_info = et.fromstring(x_str)

print(my_info)
print(type(my_info))
```

```python
# my_info 객체는 x_str에서 가장 먼저 등장하는 부모 element인 info에 해당
# my_info 객체의 tag 속성을 출력하면 element 이름 확인 가능
print(my_info.tag)
```

```python
# 자식 element 확인 방법

# 1. my_info element의 자식 element 구하기
info_childs = list(my_info)
print(info_childs)
print()

# 2. 각각의 자식 element에 담긴 text를 읽어오기
name, job, age = info_childs
print(name.text)
print(job.text)
print(age.text)
```

```python
# 자식 element가 일정하다는 보장이 없기 때문에 위와 같이 순서대로 자식 element를 찾는 것은 위험함
# findtext() method에서 찾으려는 tag를 넣어주면 해당 element를 찾아서 text로 변환
name = my_info.findtext("name")
job = my_info.findtext("job")
age = my_info.findtext("age")

print(name)
print(job)
print(age)
```

```python
# 여러 자식 element 확인
x2_str = """
<people>
    <person>
        <name>김태경1</name>
        <job>개발자</job>
        <age>20</age>
    </person>
    <person>
        <name>김태경2</name>
        <job>댄서</job>
        <age>21</age>
    </person>
</people>
"""

everyone = et.fromstring(x2_str)
print(everyone.tag) # 부모 element
print()

for one in everyone.findall("person") :
    name = one.findtext("name")
    job = one.findtext("job")
    age = one.findtext("age")
    print(name)
    print(job)
    print(age)
    print()
```

```python
# XML을 데이터프레임으로 바로 바꾸는 방법
pd.read_xml(StringIO(x2_str))
```

```python
# API key를 환경변수로 설정하고 불러오는 방법
import os
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv("API_KEY")

# API key로 URL
import requests

url = "https://data4library.kr/api/loanItemSrch?format=json&startDt=2021-04-01&endDt=2021-04-30&authKey={}".format(API_KEY)
r = requests.get(url) # url을 HTTP GET 방식으로 호출 -> requests.Response 객체 반환

print(type(r))
print(r)
```

```python
data = r.json() # requests.Response 객체의 json() method를 호출하면 json 문자열을 python 객체로 변환
print(data)

data # print() 함수로 출력하지 않아도 코드 셀의 마지막 라인의 실행 결과를 자동으로 출력
```

```python
books = []

# books = [d["doc"] for d in data["response"]["docs"]] 와 동일한 코드
for d in data["response"]["docs"] :
    books.append(d["doc"])

books
```

```python
# 도서 정보가 담긴 dictionary의 list를 데이터프레임으로 변환
books_df = pd.DataFrame(books)
books_df
```

```python
# 데이터프레임을 json으로 변환
books_df.to_json("20_best_book.json")
```
